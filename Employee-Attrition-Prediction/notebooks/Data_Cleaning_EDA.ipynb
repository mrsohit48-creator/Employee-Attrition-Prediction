```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from pathlib import Path
import warnings

warnings.filterwarnings("ignore")

sns.set_theme(style="whitegrid")

print("Libraries imported successfully!")
```

```python
PROJECT_ROOT = Path.cwd()

if PROJECT_ROOT.name == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent

DATA_PATH = (
    PROJECT_ROOT
    / "data"
    / "processed"
    / "employee_attrition_with_id.csv"
)

df = pd.read_csv(DATA_PATH)

print("Dataset loaded successfully!")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

df.head()
```

```python
print("Dataset Information:")
df.info()

print("\nStatistical Summary:")
display(df.describe())

print("\nFirst 5 Records:")
display(df.head())
```

```python
missing_values = df.isnull().sum()

missing_report = pd.DataFrame({
    "Column": missing_values.index,
    "Missing_Count": missing_values.values,
    "Missing_Percentage": (
        missing_values.values / len(df) * 100
    )
})

missing_report = missing_report.sort_values(
    by="Missing_Count",
    ascending=False
)

display(missing_report.head(15))

print("Total missing values:", df.isnull().sum().sum())
```

```python
print("Duplicate rows:", df.duplicated().sum())

print(
    "Duplicate Employee IDs:",
    df["EmployeeID"].duplicated().sum()
)
```

```python
cleaned_df = df.copy()

# Remove completely duplicated records
cleaned_df = cleaned_df.drop_duplicates()

# Remove records without EmployeeID
cleaned_df = cleaned_df.dropna(
    subset=["EmployeeID"]
)

# Convert EmployeeID to integer
cleaned_df["EmployeeID"] = (
    cleaned_df["EmployeeID"].astype(int)
)

# Remove extra spaces from text columns
text_columns = cleaned_df.select_dtypes(
    include=["object"]
).columns

for column in text_columns:
    cleaned_df[column] = (
        cleaned_df[column]
        .astype(str)
        .str.strip()
    )

# Standardize Attrition values
cleaned_df["Attrition"] = (
    cleaned_df["Attrition"].str.title()
)

print("Data cleaning completed!")
print("Original rows:", len(df))
print("Cleaned rows:", len(cleaned_df))
```

```python
attrition_counts = cleaned_df["Attrition"].value_counts()

print(attrition_counts)

attrition_rate = (
    cleaned_df["Attrition"].eq("Yes").mean() * 100
)

print(f"Overall Attrition Rate: {attrition_rate:.2f}%")
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(
    data=cleaned_df,
    x="Attrition"
)

plt.title("Employee Attrition Distribution")
plt.xlabel("Attrition")
plt.ylabel("Number of Employees")

plt.tight_layout()
plt.savefig(
    PROJECT_ROOT / "images" / "attrition_distribution.png",
    dpi=300
)

plt.show()
```

```python
department_attrition = pd.crosstab(
    cleaned_df["Department"],
    cleaned_df["Attrition"]
)

department_attrition["Total_Employees"] = (
    department_attrition.sum(axis=1)
)

department_attrition["Attrition_Rate"] = (
    department_attrition["Yes"]
    / department_attrition["Total_Employees"]
    * 100
)

display(department_attrition)

plt.figure(figsize=(10, 6))

sns.barplot(
    data=cleaned_df,
    x="Department",
    y=cleaned_df["Attrition"].eq("Yes").astype(int),
    estimator="mean"
)

plt.title("Attrition Rate by Department")
plt.xlabel("Department")
plt.ylabel("Attrition Rate")

plt.tight_layout()
plt.savefig(
    PROJECT_ROOT / "images" / "attrition_by_department.png",
    dpi=300
)
plt.show()
```

```python
overtime_attrition = pd.crosstab(
    cleaned_df["OverTime"],
    cleaned_df["Attrition"],
    normalize="index"
) * 100

display(overtime_attrition)

plt.figure(figsize=(8, 5))

sns.countplot(
    data=cleaned_df,
    x="OverTime",
    hue="Attrition"
)

plt.title("Employee Attrition Based on Overtime")
plt.xlabel("Overtime")
plt.ylabel("Number of Employees")

plt.tight_layout()
plt.savefig(
    PROJECT_ROOT / "images" / "attrition_by_overtime.png",
    dpi=300
)
plt.show()
```

```python
satisfaction_attrition = pd.crosstab(
    cleaned_df["JobSatisfaction"],
    cleaned_df["Attrition"],
    normalize="index"
) * 100

display(satisfaction_attrition)

plt.figure(figsize=(9, 5))

sns.countplot(
    data=cleaned_df,
    x="JobSatisfaction",
    hue="Attrition"
)

plt.title("Attrition by Job Satisfaction")
plt.xlabel("Job Satisfaction Score")
plt.ylabel("Number of Employees")

plt.tight_layout()
plt.savefig(
    PROJECT_ROOT / "images" / "attrition_by_satisfaction.png",
    dpi=300
)
plt.show()
```

```python
plt.figure(figsize=(9, 5))

sns.boxplot(
    data=cleaned_df,
    x="Attrition",
    y="Age"
)

plt.title("Age Distribution by Attrition")
plt.xlabel("Attrition")
plt.ylabel("Age")

plt.tight_layout()
plt.savefig(
    PROJECT_ROOT / "images" / "age_by_attrition.png",
    dpi=300
)
plt.show()
```

```python
plt.figure(figsize=(9, 5))

sns.boxplot(
    data=cleaned_df,
    x="Attrition",
    y="MonthlyIncome"
)

plt.title("Monthly Income by Attrition")
plt.xlabel("Attrition")
plt.ylabel("Monthly Income")

plt.tight_layout()
plt.savefig(
    PROJECT_ROOT / "images" / "income_by_attrition.png",
    dpi=300
)
plt.show()
```

```python
numeric_df = cleaned_df.select_dtypes(
    include=["number"]
)

correlation = numeric_df.corr()

plt.figure(figsize=(18, 12))

sns.heatmap(
    correlation,
    cmap="coolwarm",
    annot=False,
    linewidths=0.3
)

plt.title("Employee Attrition Dataset - Correlation Heatmap")

plt.tight_layout()
plt.savefig(
    PROJECT_ROOT / "images" / "correlation_heatmap.png",
    dpi=300
)
plt.show()
```

```python
total_employees = len(cleaned_df)

employees_left = (
    cleaned_df["Attrition"].eq("Yes").sum()
)

employees_stayed = (
    cleaned_df["Attrition"].eq("No").sum()
)

attrition_percentage = (
    employees_left / total_employees * 100
)

department_rates = (
    cleaned_df.assign(
        Left=cleaned_df["Attrition"].eq("Yes")
    )
    .groupby("Department")["Left"]
    .mean()
    * 100
)

highest_attrition_department = department_rates.idxmax()

overtime_rates = (
    cleaned_df.groupby("OverTime")["Attrition"]
    .apply(lambda x: x.eq("Yes").mean() * 100)
)

highest_overtime_group = overtime_rates.idxmax()

insights = f"""
EMPLOYEE ATTRITION ANALYSIS - BUSINESS INSIGHTS

1. Total Employees: {total_employees}

2. Employees Who Left: {employees_left}

3. Employees Who Stayed: {employees_stayed}

4. Overall Attrition Rate: {attrition_percentage:.2f}%

5. Department with Highest Attrition Rate:
   {highest_attrition_department}
   Attrition Rate: {department_rates.max():.2f}%

6. Overtime Group with Higher Attrition Rate:
   {highest_overtime_group}
   Attrition Rate: {overtime_rates.max():.2f}%
"""

print(insights)

reports_path = PROJECT_ROOT / "reports"
reports_path.mkdir(parents=True, exist_ok=True)

with open(
    reports_path / "eda_insights.txt",
    "w",
    encoding="utf-8"
) as file:
    file.write(insights)
```

```python
processed_path = (
    PROJECT_ROOT
    / "data"
    / "processed"
    / "employee_attrition_cleaned.csv"
)

processed_path.parent.mkdir(
    parents=True,
    exist_ok=True
)

cleaned_df.to_csv(
    processed_path,
    index=False
)

print("Cleaned dataset saved successfully!")
print("Location:", processed_path)
```

```python
quality_report = pd.DataFrame({
    "Metric": [
        "Original Rows",
        "Cleaned Rows",
        "Original Columns",
        "Missing Values",
        "Duplicate Rows",
        "Unique Employee IDs",
        "Employees Who Left",
        "Employees Who Stayed",
        "Attrition Rate"
    ],
    "Value": [
        len(df),
        len(cleaned_df),
        df.shape[1],
        cleaned_df.isnull().sum().sum(),
        cleaned_df.duplicated().sum(),
        cleaned_df["EmployeeID"].nunique(),
        employees_left,
        employees_stayed,
        f"{attrition_percentage:.2f}%"
    ]
})

quality_report.to_csv(
    reports_path / "data_quality_report_phase3.csv",
    index=False
)

display(quality_report)
```
